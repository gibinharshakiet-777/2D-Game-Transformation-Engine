```python
# ============================================================
# 🎮 GAME MATRIX — 2D GAME TRANSFORMATION ENGINE
# Engineering Mathematics Mini Project - F3
# Matrix Multiplication & Transformation Composition
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. PROJECT TITLE
# ------------------------------------------------------------

print("=" * 70)
print("                 🎮 GAME MATRIX")
print("          2D GAME TRANSFORMATION ENGINE")
print("=" * 70)

# ------------------------------------------------------------
# 2. INPUTS
# ------------------------------------------------------------

print("\nEnter transformation values:")

theta = float(input("Rotation angle (degrees): "))
scale = float(input("Scale factor: "))
tx = float(input("Translation X (tx): "))
ty = float(input("Translation Y (ty): "))

# ------------------------------------------------------------
# 3. ORIGINAL TRIANGLE
# ------------------------------------------------------------

# Triangle vertices
A = np.array([1, 1, 1])
B = np.array([3, 1, 1])
C = np.array([2, 3, 1])

triangle = np.array([A, B, C]).T

print("\nOriginal Triangle:")
print("A =", A[:2])
print("B =", B[:2])
print("C =", C[:2])

# ------------------------------------------------------------
# 4. CONVERT ANGLE TO RADIANS
# ------------------------------------------------------------

theta_rad = np.radians(theta)

# ------------------------------------------------------------
# 5. SCALING MATRIX
# ------------------------------------------------------------

S = np.array([
    [scale, 0, 0],
    [0, scale, 0],
    [0, 0, 1]
])

# ------------------------------------------------------------
# 6. ROTATION MATRIX
# ------------------------------------------------------------

R = np.array([
    [np.cos(theta_rad), -np.sin(theta_rad), 0],
    [np.sin(theta_rad),  np.cos(theta_rad), 0],
    [0, 0, 1]
])

# ------------------------------------------------------------
# 7. TRANSLATION MATRIX
# ------------------------------------------------------------

T = np.array([
    [1, 0, tx],
    [0, 1, ty],
    [0, 0, 1]
])

# ------------------------------------------------------------
# 8. DISPLAY MATRICES
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TRANSFORMATION MATRICES")
print("=" * 70)

print("\nScaling Matrix (S):")
print(np.round(S, 3))

print("\nRotation Matrix (R):")
print(np.round(R, 3))

print("\nTranslation Matrix (T):")
print(np.round(T, 3))

# ------------------------------------------------------------
# 9. STEP-BY-STEP TRANSFORMATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP-BY-STEP TRANSFORMATION")
print("=" * 70)

# Step 1: Scaling
scaled = S @ triangle

print("\n1️⃣ After Scaling:")
print(np.round(scaled, 3))

# Step 2: Rotation
rotated = R @ scaled

print("\n2️⃣ After Rotation:")
print(np.round(rotated, 3))

# Step 3: Translation
translated = T @ rotated

print("\n3️⃣ After Translation:")
print(np.round(translated, 3))

# ------------------------------------------------------------
# 10. TOTAL TRANSFORMATION MATRIX
# ------------------------------------------------------------

Total = T @ R @ S

print("\n" + "=" * 70)
print("TOTAL TRANSFORMATION MATRIX")
print("=" * 70)

print("\nTotal = T × R × S")

print("\nTotal Transformation Matrix:")
print(np.round(Total, 3))

# ------------------------------------------------------------
# 11. APPLY TOTAL MATRIX
# ------------------------------------------------------------

final_triangle = Total @ triangle

print("\nFinal Transformed Triangle:")
print("A' =", np.round(final_triangle[:2, 0], 3))
print("B' =", np.round(final_triangle[:2, 1], 3))
print("C' =", np.round(final_triangle[:2, 2], 3))

# ------------------------------------------------------------
# 12. VERIFY STEP-BY-STEP = TOTAL MATRIX
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("VERIFICATION")
print("=" * 70)

if np.allclose(translated, final_triangle):
    print("✅ Step-by-step transformation = Total Matrix transformation")
else:
    print("❌ Verification failed")

# ------------------------------------------------------------
# 13. PLOTTING FUNCTION
# ------------------------------------------------------------

def draw_triangle(points, label):
    x = list(points[0]) + [points[0][0]]
    y = list(points[1]) + [points[1][0]]

    plt.plot(x, y, marker='o', label=label)


# ------------------------------------------------------------
# 14. VISUALIZATION
# ------------------------------------------------------------

plt.figure(figsize=(10, 8))

# Original
draw_triangle(triangle, "Original Triangle")

# Scaled
draw_triangle(scaled, "After Scaling")

# Rotated
draw_triangle(rotated, "After Rotation")

# Final
draw_triangle(final_triangle, "Final Transformed Triangle")

plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)

plt.grid(True)
plt.axis("equal")

plt.title("🎮 2D Game Transformation Engine")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.legend()

plt.show()

# ------------------------------------------------------------
# 15. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PROJECT COMPLETED")
print("=" * 70)

print("""
Transformation Order:

Original
   ↓
Scaling
   ↓
Rotation
   ↓
Translation
   ↓
Final Transformed Triangle

Mathematical Model:

Total = T × R × S

Final Point = Total × Original Point
""")

print("🎮 GAME MATRIX — 2D GAME TRANSFORMATION ENGINE")
print("Matrix Mathematics + 2D Game Graphics")
```
