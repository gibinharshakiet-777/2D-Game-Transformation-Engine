```python
# ============================================================
# 🎮 GAME MATRIX — 2D GAME TRANSFORMATION ENGINE
# Engineering Mathematics Mini Project - F3
# Matrix Multiplication & Transformation Composition
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

# ------------------------------------------------------------
# 1. PROJECT TITLE
# ------------------------------------------------------------

print("=" * 70)
print("                 🎮 GAME MATRIX")
print("             2D GAME TRANSFORMATION ENGINE")
print("          Engineering Mathematics Project - F3")
print("=" * 70)

print("""
This project demonstrates how matrix multiplication is used
to transform a 2D game object by:

   • Scaling
   • Rotation
   • Translation

Mathematical Model:

        Total = Translation × Rotation × Scaling
        P' = Total × P
""")

# ------------------------------------------------------------
# 2. USER INPUTS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("                 ENTER TRANSFORMATION VALUES")
print("=" * 70)

try:
    theta = float(input("Enter rotation angle θ (degrees): "))
    scale = float(input("Enter scale factor s: "))
    tx = float(input("Enter translation X (tx): "))
    ty = float(input("Enter translation Y (ty): "))
except:
    print("Invalid input. Using default values.")
    theta = 45
    scale = 2
    tx = 3
    ty = 2

if scale <= 0:
    print("Scale factor must be positive. Using s = 1.")
    scale = 1

print("\nINPUT SUMMARY")
print("-" * 40)
print(f"Rotation angle θ : {theta}°")
print(f"Scale factor s   : {scale}")
print(f"Translation tx   : {tx}")
print(f"Translation ty   : {ty}")

# ------------------------------------------------------------
# 3. ORIGINAL TRIANGLE / GAME SPRITE
# ------------------------------------------------------------

# Homogeneous coordinates
# A = (0,0), B = (2,0), C = (1,2)

triangle = np.array([
    [0, 2, 1],
    [0, 0, 2],
    [1, 1, 1]
], dtype=float)

print("\n" + "=" * 70)
print("                 ORIGINAL GAME SPRITE")
print("=" * 70)

print("""
        A
        /\\
       /  \\
      /    \\
     B------C
""")

for i, name in enumerate(["A", "B", "C"]):
    print(f"{name} = ({triangle[0,i]:.2f}, {triangle[1,i]:.2f})")

# ------------------------------------------------------------
# 4. CREATE TRANSFORMATION MATRICES
# ------------------------------------------------------------

theta_rad = np.radians(theta)

# Scaling Matrix
S = np.array([
    [scale, 0, 0],
    [0, scale, 0],
    [0, 0, 1]
], dtype=float)

# Rotation Matrix
R = np.array([
    [np.cos(theta_rad), -np.sin(theta_rad), 0],
    [np.sin(theta_rad),  np.cos(theta_rad), 0],
    [0, 0, 1]
], dtype=float)

# Translation Matrix
T = np.array([
    [1, 0, tx],
    [0, 1, ty],
    [0, 0, 1]
], dtype=float)

# ------------------------------------------------------------
# 5. DISPLAY MATRICES
# ------------------------------------------------------------

np.set_printoptions(precision=3, suppress=True)

print("\n" + "=" * 70)
print("                 TRANSFORMATION MATRICES")
print("=" * 70)

print("\nScaling Matrix S:")
print(S)

print("\nRotation Matrix R:")
print(R)

print("\nTranslation Matrix T:")
print(T)

# ------------------------------------------------------------
# 6. STEP-BY-STEP TRANSFORMATION
# ------------------------------------------------------------

scaled = S @ triangle
rotated = R @ scaled
translated = T @ rotated

print("\n" + "=" * 70)
print("                 STEP-BY-STEP CALCULATION")
print("=" * 70)

print("\nSTEP 1 — Scaling")
print("Scaled = S × P")
print(scaled)

print("\nSTEP 2 — Rotation")
print("Rotated = R × Scaled")
print(rotated)

print("\nSTEP 3 — Translation")
print("Translated = T × Rotated")
print(translated)

# ------------------------------------------------------------
# 7. TOTAL TRANSFORMATION MATRIX
# ------------------------------------------------------------

Total = T @ R @ S

print("\n" + "=" * 70)
print("                 TOTAL TRANSFORMATION")
print("=" * 70)

print("\nTotal = T × R × S\n")
print(Total)

# Verify direct transformation
direct_result = Total @ triangle

print("\nVerification:")
print("Total × Original Triangle =")
print(direct_result)

# ------------------------------------------------------------
# 8. FINAL COORDINATES
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("                 FINAL COORDINATES")
print("=" * 70)

print(f"{'Point':<10}{'Original':<20}{'Final':<20}")
print("-" * 50)

for i, name in enumerate(["A", "B", "C"]):
    ox = triangle[0, i]
    oy = triangle[1, i]

    fx = translated[0, i]
    fy = translated[1, i]

    print(
        f"{name:<10}"
        f"({ox:.3f}, {oy:.3f}){'':<8}"
        f"({fx:.3f}, {fy:.3f})"
    )

# ------------------------------------------------------------
# 9. TRANSFORMATION ORDER COMPARISON
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("              TRANSFORMATION ORDER COMPARISON")
print("=" * 70)

# Correct order: T × R × S
result_TRS = T @ R @ S @ triangle

# Different order: S × R × T
result_SRT = S @ R @ T @ triangle

print("\nOrder 1: T × R × S")
print(result_TRS)

print("\nOrder 2: S × R × T")
print(result_SRT)

print("""
IMPORTANT:
Matrix multiplication is NOT commutative.

Therefore:

        T × R × S ≠ S × R × T

Changing the order can produce a different
final position and orientation.
""")

# ------------------------------------------------------------
# 10. STATIC VISUALIZATION
# ------------------------------------------------------------

def closed_triangle(points):
    return np.column_stack((points[:2], points[:2, 0]))

plt.figure(figsize=(11, 8))

original_plot = closed_triangle(triangle)
scaled_plot = closed_triangle(scaled)
rotated_plot = closed_triangle(rotated)
final_plot = closed_triangle(translated)

plt.plot(
    original_plot[0],
    original_plot[1],
    "o--",
    linewidth=2,
    markersize=7,
    label="Original"
)

plt.plot(
    scaled_plot[0],
    scaled_plot[1],
    "s--",
    linewidth=2,
    markersize=7,
    label="After Scaling"
)

plt.plot(
    rotated_plot[0],
    rotated_plot[1],
    "^--",
    linewidth=2,
    markersize=7,
    label="After Rotation"
)

plt.plot(
    final_plot[0],
    final_plot[1],
    "o-",
    linewidth=3,
    markersize=8,
    label="Final Sprite"
)

# Label final points
for i, name in enumerate(["A'", "B'", "C'"]):
    plt.annotate(
        name,
        (translated[0,i], translated[1,i]),
        xytext=(8,8),
        textcoords="offset points",
        fontsize=11
    )

plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)

plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")

plt.title(
    "🎮 GAME MATRIX — 2D Game Transformation Engine\n"
    f"Rotation = {theta}°, Scale = {scale}, "
    f"Translation = ({tx}, {ty})"
)

plt.grid(True)
plt.axis("equal")
plt.legend()

plt.show()

# ------------------------------------------------------------
# 11. ANIMATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("                 TRANSFORMATION ANIMATION")
print("=" * 70)

frames = 60

fig, ax = plt.subplots(figsize=(9, 7))

all_points = np.hstack([
    triangle[:2],
    translated[:2]
])

xmin = np.min(all_points[0]) - 3
xmax = np.max(all_points[0]) + 3
ymin = np.min(all_points[1]) - 3
ymax = np.max(all_points[1]) + 3

ax.set_xlim(xmin, xmax)
ax.set_ylim(ymin, ymax)

ax.axhline(0, linewidth=1)
ax.axvline(0, linewidth=1)
ax.grid(True)

ax.set_xlabel("X Coordinate")
ax.set_ylabel("Y Coordinate")

ax.set_title("🎮 GAME MATRIX — Transformation Animation")

original_closed = closed_triangle(triangle)

original_line, = ax.plot(
    original_closed[0],
    original_closed[1],
    "o--",
    linewidth=2,
    label="Original"
)

moving_line, = ax.plot(
    [],
    [],
    "o-",
    linewidth=3,
    label="Transforming Sprite"
)

ax.legend()

def interpolate_matrix(progress):

    current_scale = 1 + (scale - 1) * progress
    current_theta = theta_rad * progress
    current_tx = tx * progress
    current_ty = ty * progress

    S_current = np.array([
        [current_scale, 0, 0],
        [0, current_scale, 0],
        [0, 0, 1]
    ])

    R_current = np.array([
        [np.cos(current_theta),
         -np.sin(current_theta), 0],

        [np.sin(current_theta),
         np.cos(current_theta), 0],

        [0, 0, 1]
    ])

    T_current = np.array([
        [1, 0, current_tx],
        [0, 1, current_ty],
        [0, 0, 1]
    ])

    return T_current @ R_current @ S_current


def update(frame):

    progress = frame / (frames - 1)

    current_matrix = interpolate_matrix(progress)

    current_triangle = current_matrix @ triangle

    current_closed = closed_triangle(current_triangle)

    moving_line.set_data(
        current_closed[0],
        current_closed[1]
    )

    ax.set_title(
        f"🎮 GAME MATRIX\n"
        f"Transformation Progress: {progress*100:.0f}%"
    )

    return moving_line,


animation = FuncAnimation(
    fig,
    update,
    frames=frames,
    interval=40,
    blit=True
)

plt.close(fig)

display(HTML(animation.to_jshtml()))

# ------------------------------------------------------------
# 12. PROJECT SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("                       PROJECT SUMMARY")
print("=" * 70)

print("""
🎮 TEAM NAME:
GAME MATRIX

🎮 PROJECT:
2D Game Transformation Engine

📐 MATHEMATICS USED:
• Homogeneous Coordinates
• Matrix Representation
• Matrix Multiplication
• Transformation Composition

🔄 TRANSFORMATIONS:
• Scaling
• Rotation
• Translation

📊 MATHEMATICAL MODEL:

        P' = T × R × S × P

where:

        S = Scaling Matrix
        R = Rotation Matrix
        T = Translation Matrix
        P = Original Point
        P' = Transformed Point

🎯 REAL-WORLD APPLICATION:

2D computer graphics and games commonly use
transformation mathematics to control the position,
rotation and scale of game objects.

✅ PROJECT COMPLETE
""")

print("=" * 70)
print("                 🎮 GAME MATRIX")
print("             2D GAME TRANSFORMATION ENGINE")
print("=" * 70)
```
